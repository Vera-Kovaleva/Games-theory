## Задание 1. Метод полного перебора

```python
import numpy as np

A = np.array([
    [2, 1, 3],
    [1, 2, 1]
])

b = np.array([20, 15])
c = np.array([5, 4, 6])
max_sum = 0
for x1 in range(1, 16):
    for x2 in range(1, 16):
        for x3 in range(1, 16):
            x = [x1, x2, x3]
            if np.all(np.dot(A, x) <= b) * np.all(np.dot(x, c) > max_sum):
                max_sum = np.dot(x, c)
                res = [x1, x2, x3]
print(f"Оптимальный план: x1 = {res[0]}, x2 = {res[1]}, x3 = {res[2]}")
print(f"Максимум целевой функции: f = {max_sum}")
```

## Задание 2. Симплекс-метод

```python
import numpy as np


def simplex(A_b, A_n, C_b, C_n, b):
    basis = [3, 4]      
    nonbasis = [1, 2]   

    while True:
        delta = np.transpose(C_n) - np.dot(np.dot(np.transpose(C_b), np.linalg.inv(A_b)), A_n)
        if delta[0] > delta[1]:
            max_delta = delta[0]
            j = 0
        else:
            max_delta = delta[1]
            j = 1
        if max_delta <= 0:
            break

        a_j = A_n[:, j]
        a_s = np.dot(np.linalg.inv(A_b), a_j)
        x_b = np.dot(np.linalg.inv(A_b), b)

        theta = np.full(2, np.inf)
        theta[a_s > 0] = x_b[a_s > 0] / a_s[a_s > 0]

        index = np.argmin(theta)
        min_theta = theta[index]

        A_b_col = A_b[:, index].copy()
        A_b[:, index] = A_n[:, j]
        A_n[:, j] = A_b_col

        C_b[index], C_n[j] = C_n[j], C_b[index]
        basis[index], nonbasis[j] = nonbasis[j], basis[index]

    x_b = np.dot(np.linalg.inv(A_b), b)
    return x_b, np.dot(C_b, x_b)



C_b = np.array([0, 0])
C_n = np.array([4, 3])

A_b = np.array([
    [1, 0],
    [0, 1]
])

A_n = np.array([
    [2, 1],
    [1, 2]
])

b = np.array([8, 6])

simplex(A_b, A_n, C_b, C_n, b)              

C_b = np.array([0, 0])
C_n = np.array([4, 3])

A_b = np.array([
    [1, 0],
    [0, 1]
])

A_n = np.array([
    [2, 1],
    [1, 2]
])

b = np.array([8, 6])

x, f = simplex(A_b, A_n, C_b, C_n, b)
print(f"Оптимальный план: x1 = {x[0]:.2f}, x2 = {x[1]:.2f}")
print(f"Максимум целевой функции: f = {f:.2f}")
```
